```python
import numpy as np

a =np.random.randn(2)
b =np.random.randn(2)

anorm =np.linalg.norm(a)
bnorm =np.linalg.norm(b)

print(f"a ={a}")
print(f"b ={b}")
print(f"anorm ={anorm}")
print(f"bnorm ={bnorm}")

theta =np.arccos( np.dot(a,b) / ( anorm * bnorm ) )

print( f"theta =np.arccos( np.dot(a,b) / ( anorm * bnorm ) ) ={theta} radians" )
theta_deg =theta * (180 / np.pi);
print( f"theta_deg = theta * (180 / np.pi) = {theta_deg}" ) ;

print( "defn dot product :   dot(a,b) = |a| |b| cos(theta)" )
print( "defn dot product :   cos(theta) = dot(a,b) / (|a| |b|)" )
print( "defn dot product :   theta = arccos( dot(a,b) / (|a| |b|) )" )

u =np.array([0,1])
v =np.array([1,1])
theta_deg =(180/np.pi) * np.arccos( np.dot(u,v) / ( np.linalg.norm( u)*np.linalg.norm( v) ))
print( f"\n u ={u}\n v ={v}\n theta_deg ={theta_deg}\n" )
```

```python
mapNameToVec =dict(
   u000 =np.array([1,0])
  ,u045 =np.array([1, 1])
  ,u135 =np.array([-1,1])
  ,u225 =np.array([-1,-1])
  ,u315 =np.array([1,-1])
  ,u090 =np.array([0,1])
  ,u180 =np.array([-1,0])
  ,u270 =np.array([0,-1])
)

def print_dot_cos_theta( mapNameToVec, key1,key2, description=''):
    u =mapNameToVec[key1]
    v =mapNameToVec[key2]
    print( f"{description} : np.dot( {key1}, {key2} ) ={np.dot(u,v)} ::: cos(theta) ={np.dot(u,v)/(np.linalg.norm(u) * np.linalg.norm(v))}"  )

print_dot_cos_theta( mapNameToVec, 'u000', 'u045', 'acute' )
print_dot_cos_theta( mapNameToVec, 'u000', 'u135', 'obtuse' )
print_dot_cos_theta( mapNameToVec, 'u000', 'u225', 'obtuse' )
print_dot_cos_theta( mapNameToVec, 'u000', 'u315', 'acute' )

print ("\n")
print_dot_cos_theta( mapNameToVec, 'u000', 'u000', 'colinear -same_dir' )
print_dot_cos_theta( mapNameToVec, 'u000', 'u180', 'colinear -opposite' )

print_dot_cos_theta( mapNameToVec, 'u000', 'u090', 'orthogonal' )
print_dot_cos_theta( mapNameToVec, 'u000', 'u270', 'orthogonal' )

```

```python
# hadamard multiplication is just pair-wise multiplication
u =np.array([ 1,2,3 ])
v =np.array([ -1,-2,-3 ])

w = np.multiply( u , v )
print(" # hadamard multiplication -- pairwise multipication")
print(f"u = {str(u)}")
print(f"v = {str(v)}")
print(f"w = np.multiply( u , v ) = {str(w)}")
print(f"w = u*v = {str(u*v)}")

print(f"\n# hadamard division -- pairwise division")
y =np.divide( u, v )
print(f"y = np.division( u , v ) = {str(y)}")
print(f"y = u/v = {str(u/v)}")
```

```python
print('\n# vector\'s magnitude')
v1 = np.array([ 1, 2, 3, 4, 5, 6 ])
print( "# v1 is a six dimension vector ")
print( f"# v1 = {str(v1)}" )

# methods 1-4, just like with the regular dot product, e.g.:
vl1 = np.sqrt( sum( np.multiply(v1,v1)) ) 
print( f"mag(v1) = vl1 = np.sqrt( sum( np.multiply(v1,v1)) ) ={str(vl1)}")

# method 5: take the norm
vl2 = np.linalg.norm(v1)
print(f"mag(v1) =vl2 = np.linalg.norm(v1) ={vl2}")

vl3 = np.sqrt( np.dot( v1, v1 ))
print( f"vl3 = np.sqrt( np.dot( v1, v1 )) ={vl3}" )
```
