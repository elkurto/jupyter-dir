```python
import numpy as np

v2 = [ 2, 3 ]
v3 = [ 4, 5, 6 ]

v2t =np.transpose( v2 )
v3t =np.transpose( v3 )

print( f"v3 ={v3}" )
print( f"v3t ={v3t}" )
```

```python
v2a =[2,3] 
v2b =[1,1]

print( "concat vectors")
print( "v2a + v2b =" + str(v2a + v2b) ) 

print( "\nadd vectors" )
print( "np.add( v2a, v2b ) =" + str( np.add( v2a, v2b )) )
```
