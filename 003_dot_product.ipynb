```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
## many ways to compute the dot product

v1 = np.array([ 1, 2, 3, 4, 5, 6 ])
v2 = np.array([ 0, -4,1,  -3, 6, 5 ])
print( f"v1 ={v1}");
print( f"v2 ={v2}\n");

# method 1 - pair wise multipication, then sum
dp1 = sum( np.multiply(v1,v2) )
print( f"dp1 = sum( np.multiply(v1,v2) ) ={dp1}\n")


# method 2
dp2 = np.dot( v1,v2 )
print( f"dp2 = np.dot( v1,v2 ) ={dp2}\n")

# method 3
dp3 = np.matmul( v1,v2 ) # // dp3 == 43

# method 4
dp4 = 0  # initialize

# loop over elements
for i in range(len(v1)):
    
    # multiply corresponding element and sum
    dp4 = dp4 + v1[i]*v2[i]


print(dp1,dp2,dp3,dp4) # 43, 43, 43, 43
```
