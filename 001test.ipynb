```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# two vectors in R2
v1 = np.array([ 3, -1 ])
v2 = np.array([ 2,  4 ])

v3 = v1 + v2

print("- Adding two vectors computes the net path of the two addend vectors.\n"+
      "- The sum_vector, v3, is the diagonal of a parallogram formed by arranging\n"
      "vectors, v1 and v2, from head to tail.")
# plot them
#.        pt1,      pt2,      color, 
plt.plot([0, v1[0]],[0, v1[1]],'b',label=f"v1={str(v1)}")
plt.plot([0, v2[0]]+v1[0],[0, v2[1]]+v1[1],'r',label=f"v2={str(v2)}")
plt.plot([0, v3[0]],[0, v3[1]],'k',label='v1+v2')

plt.legend()
plt.axis('square')
plt.axis((-6, 6, -6, 6 ))
plt.grid()
plt.show()

```
