```python
import numpy as np

data = np.array([[1, 2], [3, 4], [5, 6]])
print( "data =" + str( data ) )
print( "type(data) ="+ str( type(data) ) )

print( "data.ndim    ="+ str( data.ndim ) )
print( "data.shape   ="+ str( data.shape  ) + " n rows x m cols")
print( "data.size    ="+ str( data.size   ) )
print( "data.dtype   ="+ str( data.dtype  ) )
print( "data.nbytes  ="+ str( data.nbytes ) )
```

```python
d3 =np.array( [-1, 9, 16] , dtype=complex )
print( "note in complex space sqrt(-1) = 0+1j ")
sqrt_d3 =np.sqrt( d3 ) 
sqrt_d3
```

```python
print( f"sqrt_d3.real ={str( sqrt_d3.real )}")
print( f"sqrt_d3.imag ={str( sqrt_d3.imag )}")
```

```python
# use np.linspace to fill an array element with uniformly space values in a range
x = np.linspace( -1, 1, 11 ) # create array, x, with 11 values in range [-1,1]
print( f"x ={x}\n" )

y = np.sin( np.pi * x )
y = np.round( y, decimals=4 )
print( f"y = np.sin( np.pi * x ) ={y}\n" )
```

```python
# d =[ [1,2,3],
#      [4,5,6],
#      [7,8,9] ]
#
d =np.arange( 1, 10).reshape(3,3)    
print( f"d =str{d}\n" )

print( "# sum all elements")
print( f"d.sum() = {str( d.sum() )}\n")

print( "# sum each column of matrix, d" )
print( f"d.sum(axis=0) ={str( d.sum(axis=0) )}\n")

print( "# sum each row of matrix, d" )
print( f"d.sum(axis=1) ={str( d.sum(axis=1) )}\n")


```
